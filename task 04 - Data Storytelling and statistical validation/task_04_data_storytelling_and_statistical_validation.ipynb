# Statistical Validation

## Hypothesis Testing

To validate one of the business observations, an Independent Two-Sample T-Test is performed.

### Null Hypothesis (H₀)

There is no statistically significant difference in the average sales of male and female customers.

### Alternative Hypothesis (H₁)

There is a statistically significant difference in the average sales of male and female customers.

**Significance Level (α) = 0.05**

```python
male_sales = df[df["Gender"] == "Male"]["Total_Sales"]
female_sales = df[df["Gender"] == "Female"]["Total_Sales"]

print("Male Customers :", len(male_sales))
print("Female Customers :", len(female_sales))

t_stat, p_value = ttest_ind(
    male_sales,
    female_sales,
    equal_var=False
)

print("\nT-statistic :", round(t_stat,4))
print("P-value :", round(p_value,6))
```

```python
alpha = 0.05

print("="*60)

if p_value < alpha:
    print("Decision : Reject the Null Hypothesis (H₀)")
    print("\nConclusion:")
    print("There is a statistically significant difference in the average sales of male and female customers.")
else:
    print("Decision : Fail to Reject the Null Hypothesis (H₀)")
    print("\nConclusion:")
    print("There is no statistically significant difference in the average sales of male and female customers.")

print("="*60)
```

# Statistical Interpretation

The Independent Two-Sample T-Test was performed to compare the average sales generated by male and female customers.

### Test Results

- **T-statistic:** 0.6826
- **P-value:** 0.4950
- **Significance Level (α):** 0.05

Since the p-value is greater than the significance level (0.4950 > 0.05), the Null Hypothesis (H₀) cannot be rejected.

### Business Interpretation

The analysis indicates that the observed difference in average sales between male and female customers is not statistically significant. Therefore, gender alone does not appear to be a strong factor influencing customer spending behaviour in this dataset.

This suggests that future marketing strategies should not rely solely on customer gender. Instead, the business should focus on other factors such as product categories, customer age groups, purchasing behaviour, seasonal trends, and city-wise sales performance to improve business outcomes.
