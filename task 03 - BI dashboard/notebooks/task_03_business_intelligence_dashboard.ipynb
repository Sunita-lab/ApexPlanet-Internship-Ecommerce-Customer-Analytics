### 2.1 What is a KPI?

A Key Performance Indicator (KPI) is a measurable business indicator used to evaluate how effectively an organization is achieving a specific business objective.

Unlike general metrics, KPIs are directly linked to business goals and help management monitor performance, identify problems, and make informed business decisions.

In this project, KPIs will be used to measure customer purchasing behaviour, product performance, and overall sales performance.

### 2.2 KPI vs Metric

| Metric | KPI |
|---------|-----|
| Measures any business activity. | Measures progress toward a specific business objective. |
| May or may not be important for decision-making. | Directly supports business decision-making. |
| Used for monitoring general performance. | Used for evaluating strategic business performance. |
| Example: Total Products | Example: Average Order Value |

### 2.3 Characteristics of a Good KPI

A good KPI should have the following characteristics:

- Directly aligned with a business objective.
- Measurable using available data.
- Actionable for business decision-making.
- Trackable over time to monitor trends.
- Easy to understand and communicate.

### 2.4 KPI Selection Framework

Each KPI selected for this project will be evaluated using the following framework:

1. Business Objective
   - Which business objective does the KPI support?

2. Business Rationale
   - Why is this KPI important for the business?

3. Formula
   - How is the KPI calculated?

4. Business Interpretation
   - What does a high or low value indicate?

5. Business Action
   - What decisions or actions can management take based on this KPI?

6. Limitations
   - What information does this KPI not provide?

### 4.1 KPI 1 - Total Sales

#### Business Purpose

Total Sales represents the overall revenue generated from all recorded transactions. It provides an executive-level summary of business performance and serves as the primary KPI for evaluating revenue generation.

**Total Sales = Sum of Total_Sales**

```python
# KPI 1 : Total Sales

total_sales = df["Total_Sales"].sum()

print(f"Total Sales : ₹{total_sales:,.2f}")
```

#### Limitations

- It provides no information about profitability.
- It does not explain which products, customers, or regions contributed to the revenue.

### 4.3 KPI 2 - Total Orders

##### Business purpose
Measures the number of unique orders placed by customers

##### Formula
Count of unique Order_ID

### 4.4 KPI 3 - Average Order Value (AOV)
#### Business Purpose
Measures the average revenue generated per order.

#### Formula
Total Sales / Total Orders

### 4.5 KPI 4 - Repeat Purchase Rate
#### Business Purpose
Measures the percentage of customers who placed more than one order. This helps evaluate customer retention and repeat buying behavior.

#### Formula
(Number of Customers with Repeat Purchases / Total Customers) * 100


### 4.6 KPI 5 - Average Sales per Customer
#### Business Purpose
Measures the average revenue generated per customer.

#### Formula
Total Sales / Total Customers

```python
# =========================

# Executive KPI Calculations

# =========================

# Total Sales

total_sales = df['Total_Sales'].sum()

# Total Orders

total_orders = df['Order_ID'].nunique()

# Unique Customers

total_customers = df['Customer_ID'].nunique()

# Average Order Value (AOV)

average_order_value = total_sales / total_orders

# Average Sales per Customer

average_sales_per_customer = total_sales / total_customers

# Repeat Purchase Rate

customer_order_counts = df.groupby('Customer_ID')['Order_ID'].nunique()

repeat_customers = (customer_order_counts > 1).sum()

repeat_purchase_rate = (repeat_customers / total_customers) * 100

# Display Results

print(f'Total Sales: ₹{total_sales:,.2f}')

print(f'Total Orders: {total_orders:,}')

print(f'Average Order Value (AOV): ₹{average_order_value:,.2f}')

print(f'Average Sales per Customer: ₹{average_sales_per_customer:,.2f}')

print(f'Repeat Purchase Rate: {repeat_purchase_rate:.2f}%')
```

### Business Interpretation

#### Total Sales
The business generated a total revenue of ₹139,399,439.65 from all recorded transactions.

#### Total Orders
 There are 992 unique orders, indicating the total transaction volume handled during the recorded period.

#### Average Order Value (AOV)
 Each order generated an average revenue of ₹140,523.63, suggesting that transactions in the dataset have relatively high monetary value.

#### Average Sales per Customer
 On average, each customer contributed ₹147,201.10 in revenue.

#### Repeat Purchase Rate
 Approximately 5.49% of customers placed more than one order, indicating that most customers in this dataset appear to be one-time buyers while a smaller group returned for additional purchases.

These KPIs provide a high-level executive summary of business performance and will serve as the primary metrics displayed on the interactive dashboard.

## 6. Deep-Dive Analysis: Customer Segmentation

This section performs a customer segmentation analysis using demographic attributes such as Age Group and Gender. The objective is to identify which customer segments contribute the highest revenue and derive actionable business recommendations.

```python
# Sales by Age Group
age_sales = df.groupby('Age_Group')['Total_Sales'].sum().sort_values(ascending=False)
print(age_sales)
```

### Key Findings: Age Group

1. The 36–45 age group generated the highest sales.

2. The 26–35 age group was the second-highest contributor.

3. The 18–25 age group generated the lowest sales.

4. Customers between 26 and 45 years are the primary revenue drivers.

```python
# Sales by Gender
gender_sales = df.groupby('Gender')['Total_Sales'].sum().sort_values(ascending=False)
print(gender_sales)
```

### Key Findings: Gender

1. Male customers generated slightly higher sales than female customers.

2. Both genders contribute significantly to the business.

3. Marketing campaigns should continue targeting both customer segments.

```python
# Product-wise Sales
product_sales = df.groupby('Product')['Total_Sales'].sum().sort_values(ascending=False)
print(product_sales)
```

### Key Findings: Products

3. Lower-performing products may require pricing or marketing improvements.

```python
# Category-wise Sales
category_sales = df.groupby('Category')['Total_Sales'].sum().sort_values(ascending=False)
print(category_sales)
```

### Key Findings: Categories
