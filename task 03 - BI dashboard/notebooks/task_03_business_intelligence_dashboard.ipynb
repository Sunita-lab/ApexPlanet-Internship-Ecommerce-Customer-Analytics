# Task 03 - Business Intelligence Dashboard

### 3. Business Understanding

### 3.1 Problem Statement

The dataset contains transactional sales data that captures the relationship between customers, products, and purchasing activities. The primary challenge is to understand how customer characteristics, product categories, locations, and purchasing patterns influence overall sales performance. By analyzing these interactions, the business can identify meaningful patterns, monitor sales performance, and support data-driven decision-making for future business strategies.

### 3.2 About the Dataset
This dataset represents retail sales transactions involving customers, products, and orders. It captures how different customers purchase different products over time and records the resulting sales generated from those transactions.

The dataset combines customer information, product details, order information, sales values, and geographical information, making it suitable for business intelligence, customer analytics, sales analysis, and dashboard development.

#### Unit of Analysis

The unit of analysis in this dataset is a transaction line item.

Each row represents a single product purchased as part of a customer order rather than an entire customer or a complete order.

A single Order ID may appear multiple times because one order can contain multiple products. Therefore, Order ID should be treated differently from the number of rows while performing business analysis.

#### Main Business Entities

The dataset mainly consists of three business entities:

• Customer
    - Identified using Customer ID
    - Contains demographic information such as Age, Gender and City.

• Order
    - Identified using Order ID and Order Date.
    - Represents purchasing activity performed by customers.

• Product
    - Includes Product Name, Category and Sales information.
    - Represents items purchased during customer transactions.

#### Types of Analysis Supported

This dataset supports multiple business analyses, including:

- Customer Purchasing Behaviour Analysis
- Customer Segmentation Analysis
- Product and Category Performance Analysis
- Time-based Sales Trend Analysis
- Geographical Sales Performance Analysis

#### Dataset Limitations

The dataset does not contain some important business information, including:

- Profit or Cost
- Store or Branch information
- Online/Offline purchase channel
- Customer Reviews or Ratings
- Referral Source
- Marketing Campaign data
- Customer Satisfaction or Returns

Therefore, profitability analysis, customer satisfaction analysis, and marketing effectiveness analysis cannot be performed directly using this dataset.

### 3.3 Business Objectives

1. Understand customer purchasing behavior across different customer segments.

2. Evaluate sales performance across products, categories, cities, and time periods.

3. Identify business opportunities and high-performing areas using sales data.

4. Support data-driven business decision-making through analytical insights and interactive dashboards.

### 3.4 Stakeholders

The primary stakeholders for this analysis are:

#### 1. Business Owner / CEO
- Monitor overall business performance.
- Track sales growth and customer purchasing trends.
- Make strategic business decisions based on business insights.

#### 2. Sales Manager
- Monitor sales performance across products and categories.
- Identify high-performing and low-performing products.
- Improve sales strategies using performance trends.

#### 3. Marketing Manager
- Identify target customer segments.
- Design customer-focused marketing campaigns.
- Promote products based on customer purchasing behaviour and seasonal demand.




### 3.5 Business Questions

#### Customer Purchasing Behaviour

- Which age group generates the highest sales?
- Which age group has the highest average sales per customer?
- Which gender contributes more to overall sales?
- Do loyal customers contribute significantly to total revenue?
- What is the average order value for each customer segment?

---

#### Product Performance

- Which products generate the highest sales?
- Which product categories perform best?
- Are a few products dominating sales, or is demand evenly distributed?
- Which products perform consistently over time?

---

#### Sales Performance

- Which months generate the highest sales?
- Which cities contribute the most revenue?
- Are sales trends consistent throughout the year?
- Which customer segments contribute most during peak seasons?
